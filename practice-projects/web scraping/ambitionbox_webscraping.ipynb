```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Helper function
def get_text(tag):
    return tag.text.strip() if tag else "NaN"

final = pd.DataFrame()

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/137.0.0.0 Safari/537.36"
    )
}

for j in range(1, 500):

    print(f"Scraping Page {j}")

    url = f"https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={j}"

    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        print(f"Failed Page {j}")
        continue

    soup = BeautifulSoup(response.text, "lxml")

    cards = soup.find_all("div", class_="companyCardWrapper__primaryInformation")

    company_name = []
    ratings = []
    rating_count = []
    highly_rated = []
    critical_rated = []
    location = []
    salaries = []
    interviews = []
    jobs = []
    benefits = []

    for i in cards:

        # Company Name
        company_name.append(get_text(i.find("h2")))

        # Rating
        ratings.append(get_text(i.find("div", class_="rating_star_container")))

        # Rating Count
        rating_count.append(
            get_text(
                i.find("span", class_="companyCardWrapper__companyRatingCount")
            ).strip("()")
        )

        # Location
        location.append(
            get_text(
                i.find("span", class_="companyCardWrapper__interLinking")
            )
        )

        # Highly Rated
        try:
            high = (
                i.find("span", class_="companyCardWrapper__ratingHeader--high")
                .find_parent("div")
                .find_next(
                    "span",
                    class_="companyCardWrapper__ratingValues"
                )
                .text.strip()
            )
        except:
            high = "NaN"

        highly_rated.append(high)

        # Critical Rated
        try:
            critical = (
                i.find("span", class_="companyCardWrapper__ratingHeader--critical")
                .find_parent("div")
                .find_next(
                    "span",
                    class_="companyCardWrapper__ratingValues"
                )
                .text.strip()
            )
        except:
            critical = "NaN"

        critical_rated.append(critical)

        # Salaries / Interviews / Jobs / Benefits
        counts = i.find_all(
            "span",
            class_="companyCardWrapper__ActionCount"
        )

        salaries.append(counts[1].text.strip() if len(counts) > 1 else "NaN")
        interviews.append(counts[2].text.strip() if len(counts) > 2 else "NaN")
        jobs.append(counts[3].text.strip() if len(counts) > 3 else "NaN")
        benefits.append(counts[4].text.strip() if len(counts) > 4 else "NaN")

    df = pd.DataFrame({
        "company_name": company_name,
        "ratings": ratings,
        "rating_count": rating_count,
        "highly_rated": highly_rated,
        "critical_rated": critical_rated,
        "location": location,
        "salaries": salaries,
        "interviews": interviews,
        "jobs": jobs,
        "benefits": benefits
    })

    final = pd.concat([final, df], ignore_index=True)

print(final)

# Save to CSV
final.to_csv("AmbitionBox_Companies.csv", index=False)
```
