```python
import requests
import time
import pandas as pd

session = requests.Session()
all_dfs = []

for i in range(1, 501):
    url = f'https://api.themoviedb.org/3/movie/top_rated?api_key=bd585798d974835c052460e248454770&language=en-US&page={i}'
    try:
        response = session.get(url, timeout=10)
        response.raise_for_status()
        data = response.json()
    except requests.exceptions.RequestException as e:
        print(f"Page {i} failed: {e}")
        time.sleep(2)
        continue

    if 'results' not in data:
        print(f"Page {i} bad response: {data}")
        continue

    temp_df = pd.DataFrame(data['results'])[
        ['id', 'release_date', 'title', 'original_language',
         'popularity', 'vote_average', 'vote_count']
    ]

    all_dfs.append(temp_df)

    # 👇 Progress indicator
    print(f"Completed {i}/500 pages ({i/500*100:.1f}%)", end="\r")

    time.sleep(0.2)

df = pd.concat(all_dfs, ignore_index=True)

print("\nDone!")
print(df.shape)
```
